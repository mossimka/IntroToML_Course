# Q6. Generate a random vector of length 10 and normalize it. Print the norm before and after normalization
### Answer 6

```python
v = np.random.rand(10)

norm_before = np.linalg.norm(v)

v_normalized = v / norm_before

print("Before: ", norm_before)
print("After: ", np.linalg.norm(v_normalized))
```

# Q7. Compute the dot product of vectors
a = [1,2,3,4,5,6] & b = [0,-4,3,-3,6,5]
using:
1. Manual loop
2. NumPy np.dot() .
### Answer 7

```python
a = np.array([1, 2, 3, 4, 5, 6])
b = np.array([0, -4, 3, -3, 6, 5])
manual = 0
for num1, num2 in zip(a, b):
    manual  += num1 * num2
print("Manual: ", manual)

print("numpy: ", np.dot(a, b))
```

```python
A = np.array([
[ 2, 1],
[ 5, 4],
[ 1, 6],
[ 7, 2]
])
B = np.array([
[0, 0],
[6, 3],
[3, 5]
])
print(A.shape, B.shape)
```

```python
A = A[:, np.newaxis, :]
B = B[np.newaxis, :, :]
print(A.shape, B.shape)

print(A)

print(B)
```

```python
diff = A - B
print(diff)
```

```python
sum_diff = np.sum(diff**2, axis=2) #-1
print(sum_diff)
```

```python
diff_matrix = np.sqrt(sum_diff)
print(diff_matrix)
```
