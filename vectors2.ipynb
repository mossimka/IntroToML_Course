## Vector scalar multiplication

```python
#Import numpy and Matplotlib modules
import numpy as np
import matplotlib.pyplot as plt
#ignore warning message
import warnings
warnings.filterwarnings('ignore')
```

```python
# vector and scalar
a = np.array([ 5, -3 ])
sc = 3
a_sc = a*sc # scalar-modulated
# plot them
plt.plot([0, a[0]],[0, a[1]],'k',label='a')
plt.plot([0, a_sc[0]],[0, a_sc[1]],'r--',label='$\lambda a$')
plt.legend()
plt.axis('square')
axlim = max([max(abs(a)),max(abs(a_sc))])*1.5 # dynamic axis lim
plt.axis((-axlim,axlim,-axlim,axlim))
plt.grid()
plt.show()
```

Previously, I mentioned that the data type of a variable storing a vector can sometimes be significant and
sometimes not. Vector-scalar multiplication is a case where the data type is crucial. Let us take the
following example:

```python
scal = 2
v = [3,4,5] # list
print(f'Scaler multiplication with a vector in a list is{v*scal}')

w = np.array(v) # np array
print(f'Scaler multiplication with a vector in a numpy array is{w*scal}')
```

The code first creates a scalar (scal) and a vector as a list (v), then converts that list into a NumPy array (w). In Python, the asterisk (*) has overloaded functionality; its behavior changes based on the variable type. When used with a scalar and a list, it repeats the list 'scal' times, not performing the standard scalar-vector multiplication. However, with a NumPy array, the asterisk performs elementwise multiplication. Pay attention to these differences in real-world coding, as both list repetition and vector-scalar multiplication
are common practices.
