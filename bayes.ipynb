## Calculate the probability of having a rainfall more than 500mm in month of June, provided it has flooded in that year ?

## 10- Answer this question by using Bayes Theorem and the law of the Total Probaility

P(A∣B)= P(B∣A)P(A)​ / (P(B∣A)P(A)+P(B∣¬A)P(¬A))

```python
# A = June_more_than_500
# B = FLOODS_BIN

# P(A)
P_A = len(ss[ss["June_more_than_500"] == 1]) / len(ss)

# P(not A)
P_not_A = 1 - P_A

# P(B | A)
P_B_given_A = len(ss[(ss["FLOODS_BIN"] == 1) & (ss["June_more_than_500"] == 1)]) / len(ss[ss["June_more_than_500"] == 1])

# P(B | not A)
P_B_given_not_A = len(ss[(ss["FLOODS_BIN"] == 1) & (ss["June_more_than_500"] == 0)]) / len(ss[ss["June_more_than_500"] == 0])

# Total probability: P(B) - flooded
P_B = P_B_given_A * P_A + P_B_given_not_A * P_not_A

# Bayes: P(A | B)
P_A_given_B = (P_B_given_A * P_A) / P_B

print("P(A):", P_A)
print("P(not A):", P_not_A)
print("P(B | A):", P_B_given_A)
print("P(B | not A):", P_B_given_not_A)
print("P(B):", P_B)
print("P(A | B):", P_A_given_B) # answer
```
